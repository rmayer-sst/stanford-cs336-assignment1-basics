```python
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
!uv pip install plotly 'nbformat>=4.2.0'
r = ron_rope.RoPE(max_seq_len=256,d_k=16,theta=10000)
import matplotlib.pyplot as plt
print(r.cos_tbl.shape)
import plotly.graph_objects as go

# Extract the cos_tbl and reshape it to (256, 8)
Z = r.sin_tbl[:, :]  
X, Y = np.meshgrid(range(Z.shape[0]), range(Z.shape[1]), indexing='ij')

# Create the 3D surface plot
fig = go.Figure(data=[go.Surface(z=Z.numpy(), x=X, y=Y)])

fig.update_layout(
    title="3D Plot of cos_tbl",
    scene=dict(
        xaxis_title="Sequence Length",
        yaxis_title="Feature Index",
        zaxis_title="Cosine Value"
    )
)
fig.update_layout(height=600)
fig.show()
```
